```python
while True:
    name = input("Enter your Name: ")
    id = input("Enter your ID Number: ")
    print("")

    budget = int(input(f"Good day {name}, please enter your budget:"))
    print("")
    display("Available snacks:")
    print("Pilipit = 15₱")
    print("Pinaypay = 15₱")
    print("Banana = 15₱")
    print("Banana Bread = 20₱")
    print("Siomai = 35₱")
    print("Corn = 40₱")
    print("Nutella Bread = 40₱")
    print("Corndog = 45₱")
    print("Pasta = 50₱")
    print("Sandwich = 50₱")
    print("")

    print("Enter snack: ")
    f1 = input("Snack 1 : ")
    f2 = input("Snack 2 : ")
    f3 = input("Snack 3 : ")
    f4 = input("Snack 4 : ")
    f5 = input("Snack 5 : ")
    print("")

    print("Enter how much each:")
    fa1 = int(input(f"{f1} : ")) if f1 != "n/a" else 0
    fa2 = int(input(f"{f2} : ")) if f2 != "n/a" else 0
    fa3 = int(input(f"{f3} : ")) if f3 != "n/a" else 0
    fa4 = int(input(f"{f4} : ")) if f4 != "n/a" else 0
    fa5 = int(input(f"{f5} : ")) if f5 != "n/a" else 0
    print("")

    print("Proccessing...")
    total = budget - (fa1 + fa2 + fa3 + fa4 + fa5)

    if total < 0:
        print("Out of balance")
        choice = input("Do you want to restart or end? (restart/end): ")
        if choice.lower() == 'restart':
            print("Restarting...")
            continue
        elif choice.lower() == 'end':
            print("Its ok, you can save up more money next time, please come again, thank you!")
            break
        else:
            print("Invalid choice. Ending transaction.")
            break
    else:
        print(f"Your current balance now is PHP " + str(total))
        print("Go to the counter to get your snacks and to pay, thank you and come again!")
        break
```
